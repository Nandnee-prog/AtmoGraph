```python
import torch
from torch_geometric.data import Data
from torch_geometric.nn import GCNConv

# ⚠️ SYNTHETIC / SIMULATED toy graph, sirf install test ke liye
# 5 nodes: 0=Supplier, 1=Port, 2=Manufacturer, 3=Warehouse, 4=Retailer
# Edges (direction): 0->1, 1->2, 2->3, 3->4
edge_index = torch.tensor([[0, 1, 2, 3],
                           [1, 2, 3, 4]], dtype=torch.long)

# Har node ki 2 features (jaise disruption_score, capacity), random hain
x = torch.rand(5, 2)

data = Data(x=x, edge_index=edge_index)
print(data)

# Ek GCN layer: 2 input features -> 4 output features
conv = GCNConv(in_channels=2, out_channels=4)
out = conv(data.x, data.edge_index)

print("Output shape:", out.shape)   # (5, 4) aana chahiye
```
